```python
import pandas as pd

def recomendar_restaurantes(df, categorias, estado, categorias_seleccionadas=None, promedio=None, valoraciones_min=1,
                            exclude_restaurante=None, takeout=None, delivery=None, 
                            asientos_discapacitados=None, estacionamiento_discapacitados=None,
                            almuerzo=None, cena=None, comer_solo=None, grupos=None, 
                            turistas=None, debito=None, nfc=None, credito=None,
                            servicio_rapido=None):

    # Validar el estado ingresado
    estados_validos = df['Estado'].unique()
    if estado not in estados_validos:
        return f"Estado '{estado}' no válido. Por favor, elige uno de los siguientes: {', '.join(estados_validos)}"

    # Filtrar por estado
    df_filtrado = df[df['Estado'].str.lower() == estado.lower()]

    # Si no hay restaurantes en el estado seleccionado, retornar un mensaje
    if df_filtrado.empty:
        return f"No se encontraron restaurantes en el estado '{estado}'."

    # Filtrar por categorías, si se proporcionan
    if categorias_seleccionadas:
        df_filtrado = df_filtrado[df_filtrado[['Categoria1', 'Categoria2', 'Categoria3', 'Categoria4', 'Categoria5']].apply(lambda x: x.isin(categorias_seleccionadas).any(), axis=1)]

    # Filtrar por promedio mínimo y valoraciones mínimas si se proporcionaron
    if promedio is not None:
        df_filtrado = df_filtrado[df_filtrado['Promedio'] >= promedio]
    df_filtrado = df_filtrado[df_filtrado['Valoraciones'] >= valoraciones_min]

    # Excluir restaurante si se proporciona y no es '0'
    if exclude_restaurante and exclude_restaurante != '0':
        df_filtrado = df_filtrado[df_filtrado['Nombre'].str.lower() != exclude_restaurante.lower()]

    # Calcular un puntaje de coincidencia
    df_filtrado['Puntaje'] = 0

    # Atributos a evaluar
    atributos = {
        'Takeout': takeout,
        'Delivery': delivery,
        'Asientos Discapacitados': asientos_discapacitados,
        'Estacionamiento Discapacitados': estacionamiento_discapacitados,
        'Almuerzo': almuerzo,
        'Cena': cena,
        'Comer Solo': comer_solo,
        'Grupos': grupos,
        'Turistas': turistas,
        'Debito': debito,
        'NFC': nfc,
        'Credito': credito,
        'Servicio Rapido': servicio_rapido
    }

    # Evaluar cada atributo y aumentar el puntaje
    for atributo, valor in atributos.items():
        if valor is not None:
            df_filtrado['Puntaje'] += df_filtrado[atributo].apply(lambda x: 1 if x == valor else 0)

    # Obtener recomendaciones, ordenadas por puntaje
    recomendaciones = df_filtrado.sort_values(by='Puntaje', ascending=False).drop_duplicates(subset='Nombre')

    # Si no hay recomendaciones, buscar alternativas
    if recomendaciones.empty:
        return "No se encontraron restaurantes que cumplan con los criterios."

    recomendaciones_info = []
    for index, row in recomendaciones.head(3).iterrows():
        cumplimiento = {}
        for atributo, valor in atributos.items():
            if valor is None:
                cumplimiento[atributo] = "No filtrado"
            elif row[atributo] == valor:
                cumplimiento[atributo] = "Cumple"
            else:
                cumplimiento[atributo] = "No cumple"
        recomendaciones_info.append({
            'Nombre': row['Nombre'],
            'Promedio': row['Promedio'],
            'Valoraciones': row['Valoraciones'],
            'Puntaje': row['Puntaje'],
            **cumplimiento
        })
    
    return pd.DataFrame(recomendaciones_info) if recomendaciones_info else "No se encontraron restaurantes que cumplan con los criterios."

# Asegúrate de que la función obtener_filtros siga siendo la misma, ya que esa parte está bien.

# Ejemplo de uso
# Asumiendo que ML_BASE es tu DataFrame de entrada con la información de los restaurantes
# Define ML_BASE con datos de ejemplo o reales antes de llamar a esta función
# Obtener filtros llamando a la función

obtener_filtros(ML_BASE)

```
