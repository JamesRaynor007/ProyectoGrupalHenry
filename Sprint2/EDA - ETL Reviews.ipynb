01-ALABAMA

```python
import pandas as pd
import os
import glob

# Obtener la ruta del directorio actual de trabajo
current_dir = os.getcwd()

# Construir la ruta a la carpeta que contiene los archivos JSON
json_dir = os.path.join(current_dir, 'google', 'reviews', 'review-Alabama')

# Verificar si el directorio existe
if not os.path.exists(json_dir):
    print(f"El directorio {json_dir} no existe.")
else:
    # Crear una lista para almacenar los DataFrames
    dataframes = []

    # Usar glob para encontrar todos los archivos JSON en la carpeta
    json_files = glob.glob(os.path.join(json_dir, '*.json'))

    # Verificar si se encontraron archivos JSON
    if not json_files:
        print(f"No se encontraron archivos JSON en {json_dir}.")
    else:
        # Leer cada archivo JSON y agregarlo a la lista de DataFrames
        for json_file in json_files:
            try:
                df = pd.read_json(json_file, lines=True)
                dataframes.append(df)
            except ValueError as e:
                print(f"Error al leer el archivo {json_file}: {e}")
            except Exception as e:
                print(f"Ocurrió un error inesperado al leer {json_file}: {e}")

        # Concatenar todos los DataFrames en uno solo (opcional)
        if dataframes:  # Verificar que la lista no esté vacía
            alabama = pd.concat(dataframes, ignore_index=True)
            # Mostrar el DataFrame combinado
            print(alabama)
        else:
            print("No se pudieron cargar DataFrames de los archivos JSON.")
```

```python
alabama.head()
```

Filtro cruzado con los datos limpios de metadata (queremos solo comentarios sobre los locales que consideramos restaurantes y/o afines)

```python
alabama_filtrado = alabama.merge(restaurantes.to_frame(name='gmap_id'), on='gmap_id')
```

```python
alabama_filtrado.info()
```

Transformamos a fecha la columna time

```python
# Convertir la columna 'time' a datetime
alabama_filtrado['time'] = pd.to_datetime(alabama_filtrado['time'], unit='ms')
```

```python
alabama_filtrado = alabama_filtrado[["user_id","name","time","rating","text","gmap_id"]]
```

```python
alabama_filtrado.head()
```

Evaluamos duplicados en tres conceptos claves, al mismo tiempo (se duplicó una entrada), el mismo texto y el mismo usuario. Estos tres conceptos no deberían repetirse, pudiendo un mismo usuario comentar al mismo tiempo dos cosas diferentes, o en diferentes tiempos lo mismo, pero no los tres conceptos iguales.

```python
alabama_filtrado[["time","text","user_id"]].duplicated().value_counts()
```

Una pequeña comprobación de muestra

```python
duplicados = alabama_filtrado[alabama_filtrado.duplicated(subset=["time", "text", "user_id"], keep=False)]
duplicados[duplicados["name"] == "Anthony Roberts"]
```

Se manejan los duplicados para eliminarlos y tener listo el dataset del Estado para un análisis de sentimientos posterior (tanto del Estado como de la industria completa)

```python
alabama_filtrado.drop_duplicates(subset=["time", "text", "user_id"], inplace=True)
```

```python
alabama_filtrado[["time","text","user_id"]].duplicated().value_counts()
```

Se guarda el archivo limpio

```python
import os
import pandas as pd

# Define la ruta del directorio
directorio = os.getcwd()

# Verifica si el directorio "Datos Limpios" existe y si no, lo crea
ruta_datos_limpios = os.path.join(directorio, "Datos Limpios")
if not os.path.exists(ruta_datos_limpios):
    os.makedirs(ruta_datos_limpios)

# Verifica si la carpeta "metadata" existe dentro de "Datos Limpios" y si no, la crea
ruta_reviews = os.path.join(ruta_datos_limpios, "reviews")
if not os.path.exists(ruta_reviews):
    os.makedirs(ruta_reviews)


# Guarda el DataFrame como un archivo CSV en la carpeta "metadata"
alabama_filtrado.to_csv(os.path.join(ruta_reviews, "01-alabama.csv"),index= False)
```
